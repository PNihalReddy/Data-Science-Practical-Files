**LINE** **CHART**

```python
import matplotlib.pyplot as plt
```

```python
x = [1,2,3,4,5]
y = [10,20,30,40,50]

plt.figure(figsize=(4,3)) #Chart Size
plt.plot(x, y, color='b', marker='o', linestyle='dashed',linewidth=2, markersize=12 ) # line chart generation
plt.title("Line Chart") # Line Chart
plt.xlabel("Roll Number") # X-Axis
plt.ylabel("Marks") # Y-Axis
plt.show()  #Display Chart
```

**BAR CHART**

```python
x = ['A','B','C','D','E']
y = [10,20,30,40,50]

plt.bar(x,y,color='b')
plt.title("Bar Chart")
plt.xlabel("Roll Number")
plt.ylabel("Marks")
plt.show()
```

**HISTOGRAM**


```python
import random
data = [random.randint(1, 30) for _ in range(1500)]

plt.hist(data, bins=40) # histogram chart
plt.title("Histogram Example")
plt.show()
```

**PIE PLOT**


```python
categories = ['A','B','C','D','E']
sales = [10, 20, 55, 35, 45]


plt.pie(sales, labels=categories, autopct = '%1.1f%%', startangle=90)
plt.title("Pie Chart Example")
plt.show()
```
