```python
import numpy as np
```

```python
a = np.arange(1, 21)
print("Original 1D array:")
print(a)
print("Shape:", a.shape)
```

```python
reshaped_a = a.reshape(4, 5)
print("\nReshaped 2D array (3x4):")
print(reshaped_a)
print("Shape:", reshaped_a.shape)
```

```python
transposed_a = reshaped_a.T
print("\nTransposed array :")
print(transposed_a)
print("Shape:", transposed_a.shape)
```

```python
import re
```

```python
raw_data = [
    "  Hello World!! ",
    "hello world",
    "HELLO   WORLD",
    "Python Programming...",
    "python   programming",
    "  Data Science  ",
    "DATA SCIENCE!!",
    "",
    "   "
]
print("Original Data:")
for item in raw_data:
    print(repr(item))
```

```python
def clean_text(text):

    text = text.strip()

    text = text.lower()

    text = re.sub(r'[^\w\s]', '', text)

    text = re.sub(r'\s+', ' ', text)
    return text.strip()
```

```python
cleaned_data = [clean_text(entry) for entry in raw_data]

cleaned_data = [entry for entry in cleaned_data if entry != ""]

print("\nCleaned Data (before removing duplicates):")
for item in cleaned_data:
    print(item)
```

```python
final_data = list(dict.fromkeys(cleaned_data))

print("\nFinal Cleaned & De-duplicated Data:")
for item in final_data:
    print(item)
```
