```python
import random as r

population = [1,78,6,5,2,24,3,13,44]

sample_size = 3

sample = r.sample(population,sample_size)
print(sample)
```

```python
#stratified random sampling
import random

# Create groups (strata)
group_A = [1, 2, 3, 4, 5]
group_B = [6, 7, 8, 9, 10]

# Select samples from each group
sample_A = random.sample(group_A, 2)
sample_B = random.sample(group_B, 2)

# Combine the samples
sample = sample_A + sample_B

print("Sample:", sample)
```

```python
# Cluster Sample
import random

# Create clusters
cluster_1 = [1, 2, 3, 4]
cluster_2 = [5, 6, 7, 8]
cluster_3 = [9, 10, 11, 12]

clusters = [cluster_1, cluster_2, cluster_3]

# Randomly select a cluster
selected_cluster = random.choice(clusters)

print("Selected cluster:", selected_cluster)
```
