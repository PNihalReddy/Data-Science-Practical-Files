```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
np.random.seed(42)
data = pd.DataFrame({
    'score': np.random.normal(loc=70, scale=10, size=200)
})

print(data.head())
print("\nSummary statistics:")
print(data['score'].describe())
```

```python
plt.figure(figsize=(8, 5))
sns.kdeplot(data['score'], fill=True, color='skyblue')
plt.title('KDE Plot of Score Distribution')
plt.xlabel('Score')
plt.ylabel('Density')
plt.show()
```

```python
plt.figure(figsize=(8, 5))
sns.histplot(data['score'], kde=True, color='salmon', bins=20)
plt.title('Histogram + KDE of Score Distribution')
plt.xlabel('Score')
plt.ylabel('Frequency')
plt.show()
```

```python
np.random.seed(42)
df = pd.DataFrame({
    'math_score': np.random.normal(70, 10, 200),
    'reading_score': np.random.normal(65, 12, 200),
    'writing_score': np.random.normal(68, 11, 200),
    'hours_studied': np.random.normal(5, 2, 200),
})


df['reading_score'] = df['math_score'] * 0.6 + df['reading_score'] * 0.4
df['writing_score'] = df['reading_score'] * 0.7 + df['writing_score'] * 0.3

print(df.head())
```

```python
corr_matrix = df.corr()
print("Correlation Matrix:")
print(corr_matrix)
```

```python
plt.figure(figsize=(8, 6))
sns.heatmap(corr_matrix, annot=True, cmap='coolwarm', fmt='.2f', linewidths=0.5)
plt.title('Correlation Matrix Heatmap')
plt.show()
```
